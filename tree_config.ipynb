```python
import matplotlib.pyplot as plt
import networkx as nx

def plot_binary_tree(edge,pos,labels,orange_edge):
    G = nx.DiGraph()
    G.add_edges_from(edge)

    pos = pos
    labels = labels

    # Draw edges
    edge_colors = ['orange' if edge in orange_edge else 'black' for edge in G.edges()]
    nx.draw_networkx_edges(G, pos, edge_color=edge_colors, arrows=True, width=2)

    # Draw nodes: start node light blue, others white, all with black border
    node_colors = ['grey' if node == 1 else 'white' for node in G.nodes()]
    nx.draw_networkx_nodes(
        G, pos,
        node_color=node_colors,
        edgecolors='black',
        node_size=2000,
        linewidths=2
    )

    # Draw labels
    nx.draw_networkx_labels(G, pos, labels=labels, font_size=14, font_weight='bold')

    plt.axis('off')
    plt.show()

edge = [(1, 2), (1, 3), (2, 4), (2, 5), (3, 6), (3, 7)]
pos = {1: (0, 2), 2: (-1, 1), 3: (1, 1), 4: (-1.5, 0), 5: (-0.5, 0), 6: (0.5, 0), 7: (1.5, 0)}
labels = {1: "start", 2: "2", 3: "-3", 4: "-1", 5: "-3", 6: "-1", 7: "1"}
orange_edge = [(0,1)]
plot_binary_tree(edge,pos,labels,orange_edge)
```

```python
def plot_sample_tree(edge,pos,labels,orange_edge,red_node,blue_node):
    G = nx.DiGraph()
    G.add_edges_from(edge)

    pos = pos
    labels = labels

    # Draw edges
    edge_colors = ['orange' if edge in orange_edge else 'black' for edge in G.edges()]
    nx.draw_networkx_edges(G, pos, edge_color=edge_colors, arrows=True, width=2)

    # Draw nodes: start node light blue, others white, all with black border
    node_colors = ['grey' if node == 1 else 'white' for node in G.nodes()]
    node_edge_colors = ['#F8766D' if node in red_node else '#619CFF' if node in blue_node else 'black' for node in G.nodes()]
    nx.draw_networkx_nodes(
        G, pos,
        node_color=node_colors,
        edgecolors=node_edge_colors,
        node_size=2000,
        linewidths=3
    )

    # Draw labels
    nx.draw_networkx_labels(G, pos, labels=labels, font_size=14, font_weight='bold')

    plt.axis('off')
    plt.show()

edge = [(1, 2), (1, 3), (2, 4), (2, 5), (3, 6), (3, 7)]
pos = {1: (0, 2), 2: (-1, 1), 3: (1, 1), 4: (-1.5, 0), 5: (-0.5, 0), 6: (0.5, 0), 7: (1.5, 0)}
labels = {1: "start", 2: "-1", 3: "1", 4: "4", 5: "-2", 6: "1", 7: "0"}
orange_edge = [(1,2),(2,4)]
blue_node = [2,4]
red_node = [3]

plot_sample_tree(edge,pos,labels,orange_edge,red_node,blue_node)
```
