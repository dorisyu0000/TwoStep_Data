```python
import matplotlib.pyplot as plt
import networkx as nx

def plot_binary_tree():
    G = nx.DiGraph()
    G.add_edges_from([(1, 2), (1, 3), (2, 4), (2, 5), (3, 6), (3, 7)])

    pos = {1: (0, 2), 2: (-1, 1), 3: (1, 1), 4: (-1.5, 0), 5: (-0.5, 0), 6: (0.5, 0), 7: (1.5, 0)}
    labels = {1: "Start", 2: "-1", 3: "4", 4: "+2", 5: "+3", 6: "-4", 7: "+1"}

    # Draw edges
    edge_colors = ['orange' if edge == (2, 4) else 'black' for edge in G.edges()]
    nx.draw_networkx_edges(G, pos, edge_color=edge_colors, arrows=True, width=2)

    # Draw nodes: start node light blue, others white, all with black border
    node_colors = ['#8ecae6' if node == 1 else 'white' for node in G.nodes()]
    nx.draw_networkx_nodes(
        G, pos,
        node_color=node_colors,
        edgecolors='black',
        node_size=2000,
        linewidths=2
    )

    # Draw labels
    nx.draw_networkx_labels(G, pos, labels=labels, font_size=14, font_weight='bold')

    plt.axis('off')
    plt.show()

plot_binary_tree()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Data
trees = [3, 4, 5, 6]

# RT means and standard errors
rt_means = [4839.82, 5106.31, 5281.13, 5611.28]
rt_errors = [96.42, 130.74, 202.68, 240.89]

# RT_first means and standard errors
rt_first_means = [3468.40, 3554.83, 3871.20, 4098.62]
rt_first_errors = [83.21, 126.13, 193.55, 237.19]

# RT_second means and standard errors
rt_second_means = [1371.42, 1551.48, 1409.93, 1512.66]
rt_second_errors = [34.32, 28.61, 56.83, 38.22]

# Create the plot
plt.figure(figsize=(10, 6))

# Plot lines with error bars
plt.errorbar(trees, rt_means, yerr=rt_errors, label='RT Total', marker='o', capsize=5)
plt.errorbar(trees, rt_first_means, yerr=rt_first_errors, label='RT First', marker='s', capsize=5)
plt.errorbar(trees, rt_second_means, yerr=rt_second_errors, label='RT Second', marker='^', capsize=5)

# Customize the plot
plt.xlabel('Number of paths')
plt.ylabel('Response Time (ms)')
plt.title('Response Times Across Trees')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)

# Set x-axis ticks to show only whole numbers
plt.xticks(trees)

# Show the plot
plt.show()
```
